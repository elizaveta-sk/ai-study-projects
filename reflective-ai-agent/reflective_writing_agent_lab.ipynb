# Reflection in a Research Agent

This workflow drafts an essay, critiques it, and revises it using the critique.

## Load the LLM client

```python
# Load the API key from .env and create the LLM client.
from dotenv import load_dotenv
import aisuite as ai

load_dotenv()
client = ai.Client()
```

## Step 1: Generate a draft

```python
def generate_draft(prompt: str, model: str = 'openai:gpt-4o') -> str:
    # Request a complete, well-structured first draft.
    response = client.chat.completions.create(
        model=model,
        messages=[
            {'role': 'system', 'content': 'Write a complete, well-structured essay with an introduction, body paragraphs, and a conclusion.'},
            {'role': 'user', 'content': prompt},
        ],
    )
    return response.choices[0].message.content
```

## Step 2: Reflect on the draft

```python
def reflect_on_draft(draft: str, model: str = 'openai:gpt-4o-mini') -> str:
    # Ask for constructive feedback without rewriting the essay.
    response = client.chat.completions.create(
        model=model,
        messages=[
            {'role': 'system', 'content': 'Give critical but constructive feedback in paragraph form. Address structure, clarity, argument strength, and writing style. Do not rewrite the essay.'},
            {'role': 'user', 'content': f'Essay draft to review:\n\n{draft}'},
        ],
    )
    return response.choices[0].message.content
```

## Step 3: Revise the draft

```python
def revise_draft(original_draft: str, reflection: str, model: str = 'openai:gpt-4o') -> str:
    # Apply the reflection and return only the improved essay.
    response = client.chat.completions.create(
        model=model,
        messages=[
            {'role': 'system', 'content': 'Revise the essay using the feedback. Return only the complete, improved essay; improve clarity, coherence, argument strength, and flow.'},
            {'role': 'user', 'content': f'Original draft:\n\n{original_draft}\n\nFeedback to apply:\n\n{reflection}'},
        ],
    )
    return response.choices[0].message.content
```

## Run the complete workflow

```python
# Generate a draft, reflect on it, then create the revised essay.
essay_prompt = 'Should social media platforms be regulated by the government?'

draft = generate_draft(essay_prompt)
feedback = reflect_on_draft(draft)
revised = revise_draft(draft, feedback)

print('Draft:\n', draft)
print('\nFeedback:\n', feedback)
print('\nRevised essay:\n', revised)
```
